```python
import pandas as pd


data = {
    "Student": ["Amit", "Bhavna", "Chirag", "Divya", "Esha"],
    "Maths": [85, 70, 95, 60, 88],
    "Science": [70, 88, 84, 93, 72],
    "English": [95, 90, 75, 70, 86],
    "History": [60, 65, 84, 86, 92]
}

df = pd.DataFrame(data)


df.set_index("Student", inplace=True)

print("Original DataFrame:")
print(df)


df["Average"] = df.mean(axis=1)

print("\n1. Average marks of each student:")
print(df["Average"])


print("\n2. Average marks of each subject:")
print(df[["Maths", "Science", "English", "History"]].mean())

subjects = ["Maths", "Science", "English", "History"]

print("\n3. Highest marks:")
print(df[subjects].max().max())

print("Lowest marks:")
print(df[subjects].min().min())


df["Total"] = df[subjects].sum(axis=1)

print("\n4. Student with highest total marks:")
print(df["Total"].idxmax())
print("Highest Total:", df["Total"].max())
sorted_df = df.sort_values(by="Total", ascending=False)

print("\n5. DataFrame sorted by Total marks:")
print(sorted_df)
```

```python
import pandas as pd
import numpy as np


data = {
    "Day": ["Day1", "Day2", "Day3", "Day4", "Day5", "Day6", "Day7"],
    "Temperature": [30, np.nan, 32, 31, np.nan, 29, 33]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)


print("\n1. Missing values:")
print(df.isnull())


mean_temp = df["Temperature"].mean()

df_mean = df.copy()
df_mean["Temperature"] = df_mean["Temperature"].fillna(mean_temp)

print("\n2. Missing values filled with mean:")
print(df_mean)


df_ffill = df.copy()
df_ffill["Temperature"] = df_ffill["Temperature"].ffill()

print("\n3. Missing values filled using forward fill:")
print(df_ffill)


df_drop = df.dropna()

print("\n4. DataFrame after dropping missing rows:")
print(df_drop)

print("\n5. Average temperature after cleaning:")
print(df_mean["Temperature"].mean())
```

```python
import pandas as pd


data = {
    "Product": ["Product A", "Product B", "Product C", "Product D"],
    "Jan": [120, 100, 150, 90],
    "Feb": [130, 110, 145, 100],
    "Mar": [140, 120, 160, 110],
    "Apr": [150, 130, 170, 120],
    "May": [160, 140, 180, 130],
    "Jun": [170, 150, 190, 140]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)


print("\n1. Total sales of each product:")
df["Total"] = df[["Jan", "Feb", "Mar", "Apr", "May", "Jun"]].sum(axis=1)
print(df[["Product", "Total"]])

print("\n2. Average monthly sales:")
df["Average"] = df[["Jan", "Feb", "Mar", "Apr", "May", "Jun"]].mean(axis=1)
print(df[["Product", "Average"]])

best_product = df.loc[df["Total"].idxmax(), "Product"]
print("\n3. Best-selling product:", best_product)

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
monthly_total = df[months].sum()

print("\n4. Total sales for each month:")
print(monthly_total)

best_month = monthly_total.idxmax()
print("Month with highest sales:", best_month)

df["Standard Deviation"] = df[months].std(axis=1)

print("\n5. Standard deviation of each product:")
print(df[["Product", "Standard Deviation"]])

df["Q1"] = df[["Jan", "Feb", "Mar"]].sum(axis=1)
df["Q2"] = df[["Apr", "May", "Jun"]].sum(axis=1)

pivot = df.pivot_table(
    index="Product",
    values=["Q1", "Q2"],
    aggfunc="sum"
)

print("\n6. Quarterly Sales Pivot Table:")
print(pivot)
```

```python
import numpy as np

marks = np.array([
    [85, 90, 88, 92],
    [70, 65, 72, 68],
    [95, 91, 93, 96],
    [60, 75, 70, 80],
    [88, 84, 86, 90]
])

student_average = np.mean(marks, axis=1)
print("1. Average marks of each student:")
print(student_average)

subject_average = np.mean(marks, axis=0)
print("\n2. Average marks of each subject:")
print(subject_average)


print("\n3. Highest mark:", np.max(marks))
print("Lowest mark:", np.min(marks))

student_total = np.sum(marks, axis=1)
highest_student = np.argmax(student_total)

print("\n4. Student with highest total marks:")
print("Student", highest_student + 1)
print("Total marks:", student_total[highest_student])
```

```python
import numpy as np

image = np.array([
    [100, 120, 130],
    [150, 180, 200],
    [50, 80, 100]
])

normalized = image / 255
print("1. Normalized pixel values:")
print(normalized)

bright_image = image + 30
print("\n2. Image after increasing brightness by 30:")
print(bright_image)

clipped_image = np.clip(bright_image, 0, 255)
print("\n3. Image after clipping:")
print(clipped_image)

average = np.mean(image)
print("\n4. Average pixel intensity:", average)
```

```python
import numpy as np

sales = np.array([
    [120, 130, 140, 150, 160, 170],
    [100, 110, 120, 130, 140, 150],
    [150, 145, 160, 170, 180, 190],
    [90, 100, 110, 120, 130, 140]
])

total = np.sum(sales, axis=1)
print("1. Total sales of each product:")
print(total)

average = np.mean(sales, axis=1)
print("\n2. Average monthly sales:")
print(average)

best_product = np.argmax(total) + 1
print("\n3. Best-selling product: Product", best_product)

monthly_total = np.sum(sales, axis=0)
best_month = np.argmax(monthly_total) + 1
print("\n4. Month with highest overall sales: Month", best_month)

std = np.std(sales, axis=1)
print("\n5. Standard deviation of each product:")
print(std)
```
